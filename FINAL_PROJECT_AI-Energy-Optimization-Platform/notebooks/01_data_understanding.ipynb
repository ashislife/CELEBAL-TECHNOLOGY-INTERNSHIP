```python
import pandas as pd
import os
```

```python
DATA_PATH = "../data/raw"
```

```python
os.listdir(DATA_PATH)
```

```python
daily = pd.read_csv(f"{DATA_PATH}/daily_dataset.csv")

weather_daily = pd.read_csv(f"{DATA_PATH}/weather_daily_darksky.csv")

weather_hourly = pd.read_csv(f"{DATA_PATH}/weather_hourly_darksky.csv")

households = pd.read_csv(f"{DATA_PATH}/informations_households.csv")

holidays = pd.read_csv(f"{DATA_PATH}/uk_bank_holidays.csv")

acorn = pd.read_csv(f"{DATA_PATH}/acorn_details.csv", encoding="cp1252")
```

```python
datasets = {
    "Daily": daily,
    "Weather Daily": weather_daily,
    "Weather Hourly": weather_hourly,
    "Households": households,
    "Holidays": holidays,
    "Acorn": acorn
}

for name, df in datasets.items():
    print("=" * 50)
    print(name)
    print("Shape :", df.shape)
```

```python
for name, df in datasets.items():
    print("=" * 80)
    print(name)
    print(df.columns.tolist())
```

```python
for name, df in datasets.items():
    print("=" * 80)
    print(name)
    print(df.isnull().sum())
```

```python
for name, df in datasets.items():
    print("=" * 80)
    print(name)
    print(df.dtypes)
```

```python
for name, df in datasets.items():
    print(name, ":", df.shape)
```

```python
for name, df in datasets.items():
    print("\n", name)
    print(df.columns.tolist())
```

```python
for name, df in datasets.items():
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    print("\n", name)

    if len(missing) == 0:
        print("No missing values")
    else:
        print(missing)
```
