# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus

- Add logging

## 🛠️ Step 1: Import Required Libraries

Import the required Python libraries and configure logging for tracking agent execution.

````python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
````


## 🛠️ Step 2: Create Calculator Tool

Implement a calculator tool that evaluates mathematical expressions.

````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

## 🛠️ Step 3: Create Keyword Extraction Tool

Implement a keyword extraction tool to extract important words from a sentence.

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

## 🛠️ Step 4: Create Word Count Tool

Implement an additional tool to count the total number of words in a given text.

````python
# 🛠️ TOOL 3: Word Count

def word_count(text):
    return len(text.split())
````

## 🤖 Step 5: Implement the Single Agent

Create the `agent()` function that receives the user query, identifies the intent, and routes it to the appropriate tool.


````python
def agent(query: str):
    query_lower = query.lower()

    logging.info(f"Received Query: {query}")

    try:

        # Calculator Tool
        if "calculate" in query_lower:
            expression = query_lower.replace("calculate", "").strip()

            logging.info("Routing to Calculator Tool")

            return {
                "type": "calculation",
                "result": calculator(expression)
            }

        # Keyword Tool
        elif "keyword" in query_lower:
            text = query

            if "from" in query_lower:
                text = query.split("from",1)[1].strip()

            logging.info("Routing to Keyword Extraction Tool")

            return {
                "type":"keywords",
                "result":extract_keywords(text)
            }

        # Word Count Tool
        elif "count words" in query_lower:

            text = query.replace("count words","").strip()

            logging.info("Routing to Word Count Tool")

            return {
                "type":"word_count",
                "result":word_count(text)
            }

        # General Query
        else:

            logging.info("Routing to General Response")

            return {
                "type":"general",
                "result":f"General query received: {query}"
            }

    except Exception as e:

        logging.error(str(e))

        return {
            "type":"error",
            "result":str(e)
        }
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

Display the structured JSON response generated by the agent.
